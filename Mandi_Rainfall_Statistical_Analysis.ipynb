```python
import pandas as pd

file_path = r"C:\Users\Vicky\Desktop\Rainfall.xlsx"

df = pd.read_excel(file_path)

print(df.head())
print(df.columns.tolist())
print(df.shape)
```

```python
print(df[['YEAR', 'ANN']].to_string(index=False))
```

```python
# Sort data by year
df = df.sort_values(by="YEAR").reset_index(drop=True)

# Extract annual rainfall
years = df["YEAR"]
rainfall = df["ANN"]

# Calculate monsoon and non-monsoon rainfall
df["MONSOON"] = df[["JUN", "JUL", "AUG", "SEP"]].sum(axis=1)

df["NON_MONSOON"] = df[
    ["JAN", "FEB", "MAR", "APR", "MAY",
     "OCT", "NOV", "DEC"]
].sum(axis=1)

# Check prepared data
display(df[["YEAR", "ANN", "MONSOON", "NON_MONSOON"]].head(10))
```

```python
import numpy as np
from scipy import stats

descriptive_stats = rainfall.describe()

print("DESCRIPTIVE STATISTICS")
print("----------------------")
print("Number of years:", len(rainfall))
print("Mean:", rainfall.mean())
print("Median:", rainfall.median())
print("Standard deviation:", rainfall.std())
print("Minimum:", rainfall.min())
print("Maximum:", rainfall.max())
print("Range:", rainfall.max() - rainfall.min())
print("Variance:", rainfall.var())
print("Coefficient of variation (%):",
      (rainfall.std() / rainfall.mean()) * 100)
print("Skewness:", stats.skew(rainfall))
print("Kurtosis:", stats.kurtosis(rainfall))
```

```python
wettest = df.loc[df["ANN"].idxmax()]
driest = df.loc[df["ANN"].idxmin()]

print("WETTEST YEAR")
print("Year:", wettest["YEAR"])
print("Rainfall:", wettest["ANN"], "mm")

print("\nDRIEST YEAR")
print("Year:", driest["YEAR"])
print("Rainfall:", driest["ANN"], "mm")
```

```python
import pymannkendall as mk

# Mann-Kendall test
mk_result = mk.original_test(rainfall)

print("MANN-KENDALL TREND TEST")
print("----------------------")
print("Trend:", mk_result.trend)
print("S statistic:", mk_result.s)
print("Z statistic:", mk_result.z)
print("P-value:", mk_result.p)
print("Kendall's tau:", mk_result.Tau)
print("Sen's slope:", mk_result.slope, "mm/year")
print("Intercept:", mk_result.intercept)
```

```python
from scipy.stats import linregress

regression = linregress(years, rainfall)

print("LINEAR REGRESSION RESULTS")
print("-------------------------")
print("Slope:", regression.slope, "mm/year")
print("Intercept:", regression.intercept)
print("R-squared:", regression.rvalue**2)
print("P-value:", regression.pvalue)
print("Standard error:", regression.stderr)
print("Correlation coefficient:", regression.rvalue)

# Predicted rainfall
df["PREDICTED"] = regression.intercept + regression.slope * years

# Residuals
df["RESIDUALS"] = rainfall - df["PREDICTED"]
```

```python
from scipy.stats import t

n = len(rainfall)
df_residual = n - 2
critical_t = t.ppf(0.975, df_residual)

lower_slope = regression.slope - critical_t * regression.stderr
upper_slope = regression.slope + critical_t * regression.stderr

print("95% CONFIDENCE INTERVAL FOR REGRESSION SLOPE")
print("Lower limit:", lower_slope)
print("Upper limit:", upper_slope)
```

```python
import matplotlib.pyplot as plt

# Create four rainfall periods
def assign_period(year):
    if 1985 <= year <= 1994:
        return "1985–1994"
    elif 1995 <= year <= 2004:
        return "1995–2004"
    elif 2005 <= year <= 2014:
        return "2005–2014"
    elif 2015 <= year <= 2025:
        return "2015–2025"

df["PERIOD"] = df["YEAR"].apply(assign_period)

# Arrange rainfall data by period
period_order = [
    "1985–1994",
    "1995–2004",
    "2005–2014",
    "2015–2025"
]

period_data = [
    df.loc[df["PERIOD"] == p, "ANN"]
    for p in period_order
]

# Create box plot
plt.figure(figsize=(10, 6))

plt.boxplot(
    period_data,
    patch_artist=True,
    showmeans=True,
    boxprops=dict(facecolor="skyblue", edgecolor="black"),
    medianprops=dict(color="black", linewidth=1.5),
    meanprops=dict(
        marker="D",
        markerfacecolor="#B22222",
        markeredgecolor="black",
        markersize=8
    )
)

plt.xlabel("Period", fontsize=12)
plt.ylabel("Annual Rainfall (mm)", fontsize=12)
plt.title("Annual Rainfall by Period", fontsize=14)

plt.xticks(range(1, 5), period_order)
plt.grid(axis="y", alpha=0.25)

plt.plot([], [], marker="D", color="#B22222",
         linestyle="None", label="Period mean")
plt.legend(frameon=False)

plt.tight_layout()

plt.savefig(
    "Rainfall_by_Period_600dpi.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
def assign_period(year):
    if 1985 <= year <= 1994:
        return "1985–1994"
    elif 1995 <= year <= 2004:
        return "1995–2004"
    elif 2005 <= year <= 2014:
        return "2005–2014"
    else:
        return "2015–2025"

df["PERIOD"] = df["YEAR"].apply(assign_period)

period_stats = df.groupby("PERIOD")["ANN"].agg(
    ["count", "mean", "median", "std", "min", "max"]
)

print(period_stats)
```

```python
from scipy.stats import f_oneway

groups = [
    df.loc[df["PERIOD"] == period, "ANN"]
    for period in ["1985–1994", "1995–2004",
                   "2005–2014", "2015–2025"]
]

anova_result = f_oneway(*groups)

print("ONE-WAY ANOVA")
print("F-statistic:", anova_result.statistic)
print("P-value:", anova_result.pvalue)

# Eta-squared
grand_mean = rainfall.mean()

ss_between = sum(
    len(group) * (group.mean() - grand_mean)**2
    for group in groups
)

ss_total = ((rainfall - grand_mean)**2).sum()

eta_squared = ss_between / ss_total

print("Eta-squared:", eta_squared)
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey = pairwise_tukeyhsd(
    endog=df["ANN"],
    groups=df["PERIOD"],
    alpha=0.05
)

print(tukey)
```

```python
from scipy.stats import ttest_ind

early = df.loc[df["YEAR"] <= 2004, "ANN"]
late = df.loc[df["YEAR"] >= 2005, "ANN"]

welch = ttest_ind(late, early, equal_var=False)

print("WELCH'S INDEPENDENT T-TEST")
print("--------------------------")
print("Early-period mean:", early.mean())
print("Later-period mean:", late.mean())
print("Mean difference:", late.mean() - early.mean())
print("T-statistic:", welch.statistic)
print("P-value:", welch.pvalue)

# Cohen's d using pooled standard deviation
pooled_sd = np.sqrt(
    ((len(early)-1)*early.var() +
     (len(late)-1)*late.var()) /
    (len(early)+len(late)-2)
)

cohens_d = (late.mean() - early.mean()) / pooled_sd

print("Cohen's d:", cohens_d)
```

```python
from scipy.stats import shapiro

shapiro_result = shapiro(df["RESIDUALS"])

print("Shapiro-Wilk Test")
print("W Statistic:", shapiro_result.statistic)
print("p-value:", shapiro_result.pvalue)

if shapiro_result.pvalue > 0.05:
    print("Residuals do not show significant evidence of non-normality.")
else:
    print("Residuals show significant evidence of non-normality.")
```
